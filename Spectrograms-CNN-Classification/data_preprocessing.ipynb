```python
import os
import shutil
import random

# Define the source and destination directories
src_dir = r'D:\DroneSwarmsImages\Wavelet_Plots\FlyingDronesTwo'
dest_dir = r'D:\DroneSwarmsImages\Wavelet_Plots\Scalogram_Dataset_MIX'

# Define subdirectories for training, validation, and test sets
splits = ['train', 'valid', 'test']
ratios = {'train': 0.75, 'valid': 0.20, 'test': 0.05}

# Make sure destination folders exist
for split in splits:
    for folder in ['mavic2','mavic3', 'phantom4','phantom4_mavic2','phantom4_mavic3','mavic3_mavic2']:
        os.makedirs(os.path.join(dest_dir, split, folder), exist_ok=True)

# Function to rename images using a proper format
def rename_image(folder_name, image_name, idx):
    ext = os.path.splitext(image_name)[1]  # Get the file extension
    new_name = f"{folder_name}_img_{idx:04d}{ext}"  # Rename using a consistent format
    return new_name

# Process each folder and split images accordingly
for folder in  ['mavic2','mavic3', 'phantom4','phantom4_mavic2','phantom4_mavic3','mavic3_mavic2']:
    folder_path = os.path.join(src_dir, folder)
    images = os.listdir(folder_path)
    
    # Shuffle images to ensure randomness
    random.shuffle(images)
    
    # Calculate the number of images for each split
    total_images = len(images)
    train_count = int(total_images * ratios['train'])
    valid_count = int(total_images * ratios['valid'])
    test_count = total_images - train_count - valid_count
    
    # Split the images
    split_data = {
        'train': images[:train_count],
        'valid': images[train_count:train_count + valid_count],
        'test': images[train_count + valid_count:]
    }

    # Move and rename the images
    for split, img_list in split_data.items():
        for idx, img_name in enumerate(img_list):
            src_img_path = os.path.join(folder_path, img_name)
            new_img_name = rename_image(folder, img_name, idx)
            dest_img_path = os.path.join(dest_dir, split, folder, new_img_name)
            shutil.copy(src_img_path, dest_img_path)

print("Image distribution complete!")
```
